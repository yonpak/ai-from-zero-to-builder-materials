# Transformer Families Lab

Compare encoder-only, decoder-only, and encoder-decoder information access before thinking about model size or training quality.


```python
def full_visibility(length):
    return [[1 for _ in range(length)] for _ in range(length)]

def causal_visibility(length):
    return [[1 if column <= row else 0 for column in range(length)] for row in range(length)]

def cross_visibility(target_length, source_length):
    return [[1 for _ in range(source_length)] for _ in range(target_length)]

def show(name, matrix):
    print(name)
    for row in matrix:
        print(" ".join(map(str, row)))
    print()

input_length = 4
source_length = 2
target_length = 4
focus_position = 2

encoder = full_visibility(input_length)
decoder = causal_visibility(target_length)
encoder_decoder_target = causal_visibility(target_length)
encoder_decoder_cross = cross_visibility(target_length, source_length)

show("encoder-only self-attention", encoder)
show("decoder-only self-attention", decoder)
show("encoder-decoder target self-attention", encoder_decoder_target)
show("encoder-decoder cross-attention to source", encoder_decoder_cross)

assert all(value == 1 for row in encoder for value in row)
assert all(decoder[row][column] == (1 if column <= row else 0) for row in range(target_length) for column in range(target_length))
assert all(value == 1 for row in encoder_decoder_cross for value in row)

visible_decoder_positions = [i for i, value in enumerate(decoder[focus_position]) if value]
print("decoder positions visible from focus", focus_position, ":", visible_decoder_positions)
assert visible_decoder_positions == list(range(focus_position + 1))
print("PASS: l05-14 family visibility checks")
```

```python
tokens = ["the", "cat", "sat", "down"]
masked_position = 1
masked_input = tokens.copy()
masked_target = masked_input[masked_position]
masked_input[masked_position] = "[MASK]"

next_token_questions = [
    (" ".join(tokens[:i]), tokens[i])
    for i in range(1, len(tokens))
]

print("masked-token question:")
print("  input :", " ".join(masked_input))
print("  target:", masked_target)
print()
print("next-token questions:")
for prefix, target in next_token_questions:
    print(f"  prefix={prefix!r:18s} target={target!r}")

assert masked_target == "cat"
assert next_token_questions[-1] == ("the cat sat", "down")
print("PASS: l05-14 objective-structure checks")
```
