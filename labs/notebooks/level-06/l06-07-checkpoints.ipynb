# Checkpoints and Reproducibility
Create a deterministic metadata record and fingerprint it. Real model/optimizer state belongs beside the same metadata.


```python
import hashlib, json

record = {
  'run_id': 'l06-demo', 'step': 60, 'seed': 7,
  'config': {'block_size': 32, 'n_embd': 48, 'n_head': 4, 'n_layer': 2},
  'tokenizer_revision': 'toy-char-v1', 'data_split': '90-10-tail',
  'validation_loss': 2.70, 'python': '>=3.11'
}
payload = json.dumps(record, sort_keys=True).encode()
fingerprint = hashlib.sha256(payload).hexdigest()[:16]
print(json.dumps(record, indent=2, sort_keys=True))
print('fingerprint:', fingerprint)
assert fingerprint == hashlib.sha256(payload).hexdigest()[:16]
print('PASS: checkpoint metadata is deterministic and fingerprinted')
```
