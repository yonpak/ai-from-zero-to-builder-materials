# Evaluate a Tiny Language Model
Combine held-out loss with fixed generated-sample evidence.


```python
import math

records = [
 {'step':40,'val_loss':2.91,'sample':'the cat sat near the fire'},
 {'step':60,'val_loss':2.70,'sample':'the cat learned a small new pattern'},
 {'step':80,'val_loss':2.86,'sample':'the cat the cat the cat the cat'},
]
for r in records:
    r['perplexity'] = math.exp(r['val_loss'])
best = min(records, key=lambda r:r['val_loss'])
for r in records: print(r['step'], round(r['val_loss'],2), round(r['perplexity'],2), r['sample'])
assert best['step'] == 60
assert math.exp(2.70) < math.exp(2.91)
print('PASS: fixed evaluation selects step 60 by validation loss')
```
