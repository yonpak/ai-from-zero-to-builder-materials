# Validation Loss and Overfitting
Select the best checkpoint using held-out loss and identify a simple divergence pattern.


```python
steps = [0,20,40,60,80]
train = [4.2,3.1,2.4,1.9,1.5]
valid = [4.3,3.3,2.8,2.9,3.2]
best_i = min(range(len(valid)), key=valid.__getitem__)
print('best validation step:', steps[best_i])
print('best validation loss:', valid[best_i])
assert steps[best_i] == 40
assert train[-1] < train[best_i] and valid[-1] > valid[best_i]
print('PASS: later train improvement can coexist with worse validation')
```
