# Model Configuration and Shapes
Validate head width and estimate a few parameter tables before assembling the model.


```python
def inspect_config(vocab_size, block_size, n_embd, n_head):
    if n_embd % n_head != 0:
        raise ValueError('n_embd must divide evenly by n_head')
    return {
        'head_dim': n_embd // n_head,
        'token_embedding_params': vocab_size * n_embd,
        'position_embedding_params': block_size * n_embd,
        'lm_head_params': n_embd * vocab_size,
    }

report = inspect_config(64, 32, 48, 4)
print(report)
assert report['head_dim'] == 12
try:
    inspect_config(64, 32, 50, 8)
    raise AssertionError('invalid config should fail')
except ValueError:
    pass
print('PASS: shape contract rejects uneven attention heads')
```
