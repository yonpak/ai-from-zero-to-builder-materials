# Next-Token Prediction
Create one-token-shifted input/target pairs and inspect a tiny cross-entropy calculation.


```python
import math

def shifted(ids):
    if len(ids) < 2:
        raise ValueError('need at least two token ids')
    # TODO: return input IDs and next-token targets shifted by exactly one position.
    return ids, ids

x, y = shifted([7, 4, 9, 2])
print('input :', x)
print('target:', y)
assert x == [7, 4, 9] and y == [4, 9, 2]

logits = [2.0, 1.0, 0.0]
target = 0
den = sum(math.exp(v) for v in logits)
loss = -math.log(math.exp(logits[target]) / den)
print('example cross-entropy:', round(loss, 4))
assert loss > 0
print('PASS: targets are shifted by exactly one token')
```
