# Sampling Text — Controlled Toy Distribution
Keep probabilities and seed fixed so a sampling change can be reproduced.


```python
import random

tokens = ['A','B','C']
weights = [0.6,0.3,0.1]
rng = random.Random(7)
draws = rng.choices(tokens, weights=weights, k=12)
print(' '.join(draws))
rng2 = random.Random(7)
assert draws == rng2.choices(tokens, weights=weights, k=12)
assert set(draws) <= set(tokens)
print('PASS: fixed seed and distribution reproduce the sample')
```
