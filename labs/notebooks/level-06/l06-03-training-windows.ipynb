# Training Examples from Text
Split source tokens first, then make `block_size + 1` slices inside each split.


```python
def windows(ids, block_size):
    if block_size < 1 or len(ids) <= block_size:
        raise ValueError('need more tokens than block_size')
    out = []
    for start in range(len(ids) - block_size):
        source = ids[start:start + block_size + 1]
        out.append((source[:-1], source[1:]))
    return out

tokens = list(range(12))
train, valid = tokens[:9], tokens[9:]
train_examples = windows(train, 4)
print(train_examples[:2])
assert train_examples[0] == ([0,1,2,3],[1,2,3,4])
assert all(max(x) < 9 and max(y) < 9 for x, y in train_examples)
print('PASS: windows stay inside the training split')
```
