# Decoder-Only Language Models
Build a causal visibility table. A `1` means the row is allowed to use that column; a `0` means it is blocked because it is in the future.


```python
def causal_access(length):
    return [[j <= i for j in range(length)] for i in range(length)]

length = 5
mask = causal_access(length)
for row in mask:
    print(' '.join('1' if x else '0' for x in row))

for i, row in enumerate(mask):
    assert all(not allowed for j, allowed in enumerate(row) if j > i)
assert sum(sum(row) for row in mask) == length * (length + 1) // 2
print('PASS: causal access only uses the visible prefix')
```
