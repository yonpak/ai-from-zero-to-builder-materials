# Temperature, Top-k, and Top-p
Apply decoding filters to one fixed logit vector.


```python
import math

def softmax(logits, temperature=1.0):
    if temperature <= 0: raise ValueError('temperature must be positive')
    scaled = [x/temperature for x in logits]
    m = max(scaled); ex = [math.exp(x-m) for x in scaled]; z=sum(ex)
    return [x/z for x in ex]

def top_k(probs, k):
    keep = set(sorted(range(len(probs)), key=probs.__getitem__, reverse=True)[:k])
    total = sum(probs[i] for i in keep)
    return [probs[i]/total if i in keep else 0.0 for i in range(len(probs))]

def top_p(probs, p):
    # TODO: keep the smallest descending-probability prefix whose cumulative mass reaches p,
    # then renormalize the kept probabilities.
    return list(probs)

p1 = softmax([2.0,1.0,0.2,-0.5])
print('base :', [round(x,3) for x in p1])
print('top-k:', [round(x,3) for x in top_k(p1,2)])
print('top-p:', [round(x,3) for x in top_p(p1,0.8)])
assert sum(x>0 for x in top_k(p1,2)) == 2
assert abs(sum(top_p(p1,0.8))-1) < 1e-9
assert sum(x > 0 for x in top_p(p1,0.8)) == 2
print('PASS: decoding controls keep a normalized candidate distribution')
```
