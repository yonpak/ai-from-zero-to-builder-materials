# Generation Failure Modes
Compute a repeated-bigram indicator and compare a known loop with a more varied sample.


```python
def repeated_bigram_rate(text):
    words=text.split(); pairs=list(zip(words, words[1:]))
    if not pairs: return 0.0
    return 1 - len(set(pairs))/len(pairs)

loop = 'the cat the cat the cat the cat'
varied = 'the cat sat near a warm window'
loop_rate = repeated_bigram_rate(loop)
varied_rate = repeated_bigram_rate(varied)
print('loop repetition :', round(loop_rate,3))
print('varied repetition:', round(varied_rate,3))
assert loop_rate > varied_rate
print('PASS: the indicator detects the intentional repetition failure')
```
