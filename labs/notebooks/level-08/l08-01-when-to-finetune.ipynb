# When to Fine-Tune
Compare prompting, retrieval, deterministic code, and fine-tuning from explicit scenario evidence.


```python
def choose_intervention(case):
    if case["hard_rule"]:
        return "application_logic"
    if case["fresh_external_knowledge"]:
        return "retrieval"
    if case["prompt_works"]:
        return "prompting"
    if case["repeated_behavior_gap"] and case["training_examples"] and case["heldout_eval"]:
        return "fine_tuning"
    return "insufficient_evidence"

cases = [
    {"name":"today_policy","hard_rule":False,"fresh_external_knowledge":True,"prompt_works":False,"repeated_behavior_gap":False,"training_examples":False,"heldout_eval":True},
    {"name":"refund_limit","hard_rule":True,"fresh_external_knowledge":False,"prompt_works":False,"repeated_behavior_gap":True,"training_examples":True,"heldout_eval":True},
    {"name":"domain_style","hard_rule":False,"fresh_external_knowledge":False,"prompt_works":False,"repeated_behavior_gap":True,"training_examples":True,"heldout_eval":True},
]
for case in cases:
    print(case["name"], "->", choose_intervention(case))
assert choose_intervention(cases[0]) == "retrieval"
assert choose_intervention(cases[1]) == "application_logic"
assert choose_intervention(cases[2]) == "fine_tuning"
print("PASS: adaptation choices follow the declared evidence")
```
