# Build a LoRA Adapter
Implement the low-rank branch and verify zero-scale/base-preservation invariants.


```python
def matvec(matrix, vector):
    return [sum(w * x for w, x in zip(row, vector)) for row in matrix]

def add(a, b):
    return [x + y for x, y in zip(a, b)]

def lora_update(x, A, B, scale):
    # TODO: compute scale * B(Ax).
    return [0.0 for _ in B]

def adapted_linear(x, W, A, B, scale):
    base = matvec(W, x)
    update = lora_update(x, A, B, scale)
    return add(base, update)

x = [2.0, -1.0]
W = [[1.0, 0.0], [0.0, 1.0]]
A = [[0.5, 0.5]]
B = [[1.0], [-2.0]]

out = adapted_linear(x, W, A, B, 0.2)
print("adapted output:", out)
assert all(abs(a-b) < 1e-9 for a,b in zip(out, [2.1, -1.2]))

base_only = adapted_linear(x, W, A, B, 0.0)
assert base_only == [2.0, -1.0]
print("PASS: LoRA adds a scaled low-rank update without replacing the base path")
```
