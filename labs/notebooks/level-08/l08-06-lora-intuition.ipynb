# LoRA Intuition
Trace A/B shapes, parameter counts, and one tiny low-rank matrix product.


```python
def lora_shapes(out_features, in_features, rank):
    return (rank, in_features), (out_features, rank)

def lora_parameters(out_features, in_features, rank):
    return rank * in_features + out_features * rank

def matmul(a, b):
    rows, inner, cols = len(a), len(b), len(b[0])
    assert all(len(row) == inner for row in a)
    return [[sum(a[i][k] * b[k][j] for k in range(inner)) for j in range(cols)] for i in range(rows)]

A = [[0.5, 0.5]]
B = [[1.0], [-2.0]]
delta = matmul(B, A)
print("A shape:", lora_shapes(2, 2, 1)[0])
print("B shape:", lora_shapes(2, 2, 1)[1])
print("B@A:", delta)
print("params rank1:", lora_parameters(2, 2, 1))
assert delta == [[0.5, 0.5], [-1.0, -1.0]]
assert lora_parameters(4096, 4096, 8) == 65_536
print("PASS: low-rank factors restore the base matrix shape")
```
