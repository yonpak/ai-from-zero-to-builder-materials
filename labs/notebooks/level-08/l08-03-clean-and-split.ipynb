# Cleaning and Formatting Training Data
Complete a group-safe split so related turns cannot leak across train and validation.


```python
rows = [
    {"id":"a1","case_id":"A","text":"turn one"},
    {"id":"a2","case_id":"A","text":"turn two"},
    {"id":"b1","case_id":"B","text":"turn one"},
    {"id":"c1","case_id":"C","text":"turn one"},
    {"id":"c2","case_id":"C","text":"turn two"},
]

def group_split(rows, validation_groups):
    # TODO: return train_rows, validation_rows by assigning whole case_id groups.
    return rows[:3], rows[3:]

train_rows, val_rows = group_split(rows, {"C"})
train_groups = {r["case_id"] for r in train_rows}
val_groups = {r["case_id"] for r in val_rows}
print("train groups:", train_groups)
print("validation groups:", val_groups)
assert train_groups.isdisjoint(val_groups)
assert val_groups == {"C"}
print("PASS: related records stay inside one split")
```
