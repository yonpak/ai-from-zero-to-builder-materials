# Supervised Fine-Tuning
Implement a response-only masked loss over token-level losses.


```python
token_losses = [1.2, 0.9, 0.8, 0.7, 0.6, 0.5, 0.6, 0.2]
loss_mask    = [0,   0,   0,   0,   0,   0,   1,   1]

def masked_mean(losses, mask):
    # TODO: average losses only where the corresponding mask value is 1.
    return sum(losses) / len(losses)

loss = masked_mean(token_losses, loss_mask)
print("response-only loss:", loss)
assert abs(loss - 0.4) < 1e-9

changed_prompt_loss = list(token_losses)
changed_prompt_loss[1] = 9.0
assert abs(masked_mean(changed_prompt_loss, loss_mask) - loss) < 1e-9
print("PASS: masked prompt-token losses do not change the response-only objective")
```
