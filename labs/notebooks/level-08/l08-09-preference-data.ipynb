# Preference Data
Audit context consistency, ambiguity, and superficial length shortcuts in pairwise preference examples.


```python
pairs = [
    {"id":"p1","prompt":"Summarize incident","chosen":"Impact: checkout unavailable for 11 minutes.","rejected":"Database crashed for 11 minutes.","uncertain":False},
    {"id":"p2","prompt":"State battery duration","chosen":"Battery duration is 10 hours.","rejected":"It lasts ten hours.","uncertain":True},
]

def audit(pairs):
    lengths = []
    uncertain = []
    for pair in pairs:
        lengths.append((len(pair["chosen"].split()), len(pair["rejected"].split())))
        if pair["uncertain"]:
            uncertain.append(pair["id"])
    avg_chosen = sum(a for a,_ in lengths) / len(lengths)
    avg_rejected = sum(b for _,b in lengths) / len(lengths)
    return {"uncertain_ids": uncertain, "avg_chosen_words": avg_chosen, "avg_rejected_words": avg_rejected}

report = audit(pairs)
print(report)
assert report["uncertain_ids"] == ["p2"]
print("PASS: ambiguous preference evidence remains visible")
```
