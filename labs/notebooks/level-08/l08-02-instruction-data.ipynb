# Instruction-Tuning Data
Inspect which behaviors a small example set actually covers.


```python
examples = [
    {"id":"a","input":"battery drains after update","target":"battery_drain","has_evidence":True,"boundary":False},
    {"id":"b","input":"device will not charge","target":"charging_failure","has_evidence":True,"boundary":False},
    {"id":"c","input":"battery issue, details missing","target":"abstain","has_evidence":False,"boundary":True},
]

def coverage(rows):
    labels = {}
    for row in rows:
        labels[row["target"]] = labels.get(row["target"], 0) + 1
    return {
        "rows": len(rows),
        "labels": labels,
        "abstention_cases": sum(not r["has_evidence"] for r in rows),
        "boundary_cases": sum(r["boundary"] for r in rows),
    }

report = coverage(examples)
print(report)
assert report["abstention_cases"] >= 1
assert report["boundary_cases"] >= 1
print("PASS: data report exposes behavior coverage")
```
